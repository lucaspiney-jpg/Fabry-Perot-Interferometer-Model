```python
import numpy as np
import matplotlib.pyplot as plt

fig = plt.figure()
ax = fig.add_subplot(projection='3d')

radius = 3
theta_arr = np.linspace(0, 2*np.pi, 201)
y_circle = radius * np.cos(theta_arr)
z_circle = radius * np.sin(theta_arr)

plate_1_x = 5
separation = 1
plate_2_x = plate_1_x + separation

# plot the two circular plates
ax.plot(np.full_like(theta_arr, plate_1_x), y_circle, z_circle)
ax.plot(np.full_like(theta_arr, plate_2_x), y_circle, z_circle)

def reflection(x, y, z, theta, phi):
    rho = np.linspace(0, 5, 1000)

    midpoint = plate_1_x + separation/2

    # moving to the right
    if x < midpoint:
        x_light = x + rho * np.cos(theta) * np.sin(phi)
        y_light = y + rho * np.sin(theta) * np.sin(phi)
        z_light = z + rho * np.cos(phi)

        mask = x_light <= plate_2_x
        ax.plot(x_light[mask], y_light[mask], z_light[mask])

        idx = np.where(mask)[0][-1]
        
        return x_light[idx], y_light[idx], z_light[idx]

    # moving to the left
    else:
        x_light = x - rho * np.cos(theta) * np.sin(phi)
        y_light = y + rho * np.sin(theta) * np.sin(phi)
        z_light = z + rho * np.cos(phi)

        mask = x_light >= plate_1_x
        ax.plot(x_light[mask], y_light[mask], z_light[mask])

        idx = np.where(mask)[0][-1]
        return x_light[idx], y_light[idx], z_light[idx]

def fp(y0, z0, theta, phi, n_reflections=6):
    # source point
    x0 = 0
    ax.scatter(x0, y0, z0, s=40)

    rho = np.linspace(0, 10, 10000)

    x_light = rho * np.cos(theta) * np.sin(phi)
    y_light = y0 + rho * np.sin(theta) * np.sin(phi)
    z_light = z0 + rho * np.cos(phi)

    # first segment up to first plate
    mask = x_light <= plate_1_x
    ax.plot(x_light[mask], y_light[mask], z_light[mask])

    idx = np.where(mask)[0][-1]
    x_ref = x_light[idx]
    y_ref = y_light[idx]
    z_ref = z_light[idx]

    # repeated reflections
    for _ in range(n_reflections):
        x_ref, y_ref, z_ref = reflection(x_ref, y_ref, z_ref, theta, phi)
        rho_out = np.linspace(0, 10, 1000)

        x_out = x_ref + rho_out * np.cos(theta) * np.sin(phi)
        y_out = y_ref + rho_out * np.sin(theta) * np.sin(phi)
        z_out = z_ref + rho_out * np.cos(phi)

        ax.plot(x_out, y_out, z_out, '--', alpha=0.7, color = 'red')
    

fp(0, 0, 0.1, np.pi/2)
#fp(2, -0.5, 0.1, np.pi/2 - 0.1)

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")

#ax.view_init(0,0,0)
```

```python
def fabry_perot_rings(lam, d=50e-6, n=1.0, R=0.85, f=0.2,
                     N=600, screen_size=0.02):
    
    # finesse coefficient
    F = 4 * R / (1 - R)**2

    # detector grid
    x = np.linspace(-screen_size, screen_size, N)
    y = np.linspace(-screen_size, screen_size, N)
    X, Y = np.meshgrid(x, y)

    rho = np.sqrt(X**2 + Y**2)

    # angle
    theta = np.arctan(rho / f)

    # phase
    delta = (4 * np.pi * n * d * np.cos(theta)) / lam

    # intensity
    I = 1 / (1 + F * np.sin(delta / 2)**2)

    return I

# example wavelengths (nm → m)
wavelengths = np.linspace(500e-9, 700e-9, 5)

I_total = 0

for lam in wavelengths:
    I_total += fabry_perot_rings(lam)

# normalise
I_total /= len(wavelengths)

plt.figure(figsize=(8,8))
plt.imshow(I_total, cmap='gray', origin='lower')
plt.title("Fabry–Perot Rings (Multiple Wavelengths)")
plt.gca().set_aspect('equal')
plt.colorbar(label="Intensity")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt



def plot_fp_3d_physical(points, n_bounces=8,
                        wavelength=600e-9,   # metres
                        reflectivity=0.85,   # intensity reflectivity R
                        refractive_index=2,
                        mirror1_x=4.0, mirror2_x=6.0,
                        lens_x=9.0, screen_x=12.0,
                        mirror_radius=2.5, lens_radius=2.0, detector_radius=3.0,
                        figsize=(11, 8)):
    """
    3D Fabry–Perot interferometer with multiple rays and basic physical weighting.

    Each item in points should be:
        {"pos": (x0, y0, z0), "theta": ..., "phi": ...}

    Physics included:
    - wavelength-dependent phase
    - reflectivity-dependent transmitted intensity
    - detector point from lens focal mapping
    - optical path difference for each transmitted beam

    Parameters
    ----------
    wavelength : float
        Wavelength in metres.
    reflectivity : float
        Intensity reflectivity R of each mirror (0 < R < 1).
    refractive_index : float
        Refractive index inside cavity.
    """

    fig = plt.figure(figsize=figsize)
    ax = fig.add_subplot(111, projection='3d')

    # basic FP quantities
    R = reflectivity
    T = 1 - R
    d = mirror2_x - mirror1_x
    f = screen_x - lens_x

    # -------------------------
    # Optical element outlines
    # -------------------------
    t = np.linspace(0, 2*np.pi, 300)

    y_circle = mirror_radius * np.cos(t)
    z_circle = mirror_radius * np.sin(t)

    ax.plot(np.full_like(t, mirror1_x), y_circle, z_circle, lw=2.5, color='black')
    ax.plot(np.full_like(t, mirror2_x), y_circle, z_circle, lw=2.5, color='black')

    yl = np.linspace(-lens_radius, lens_radius, 300)
    zl = np.sqrt(np.clip(lens_radius**2 - yl**2, 0, None))
    ax.plot(np.full_like(yl, lens_x), yl,  zl, lw=2, alpha=0.8, color='black')
    ax.plot(np.full_like(yl, lens_x), yl, -zl, lw=2, alpha=0.8, color='black')

    yd = detector_radius * np.cos(t)
    zd = detector_radius * np.sin(t)
    ax.plot(np.full_like(t, screen_x), yd, zd, lw=2.5, color='black')

    # optical axis
    ax.plot([0, screen_x], [0, 0], [0, 0], 'k--', alpha=0.25)

    # -------------------------
    # Helpers
    # -------------------------
    def direction_vector(theta, phi, sign_x=+1):
        dx = sign_x * abs(np.cos(theta) * np.sin(phi))
        dy = np.sin(theta) * np.sin(phi)
        dz = np.cos(phi)
        return dx, dy, dz

    def propagate_to_plane(x, y, z, dx, dy, dz, x_plane, npts=400):
        s_hit = (x_plane - x) / dx
        s = np.linspace(0, s_hit, npts)

        xp = x + dx * s
        yp = y + dy * s
        zp = z + dz * s

        return xp, yp, zp, xp[-1], yp[-1], zp[-1]

    def inside_circular_aperture(y, z, radius):
        return y**2 + z**2 <= radius**2

    def clip_alpha(val, amin=0.08, amax=1.0):
        return max(amin, min(amax, val))

    colours = plt.rcParams['axes.prop_cycle'].by_key()['color']

    # -------------------------
    # Trace each ray
    # -------------------------
    for idx, ray in enumerate(points):
        c_main = colours[idx % len(colours)]

        x0, y0, z0 = ray["pos"]
        theta = ray["theta"]
        phi = ray["phi"]

        ax.scatter(x0, y0, z0, s=45, color=c_main)

        # initial ray from source to mirror 1
        dx, dy, dz = direction_vector(theta, phi, sign_x=+1)
        x_seg, y_seg, z_seg, x_hit, y_hit, z_hit = propagate_to_plane(
            x0, y0, z0, dx, dy, dz, mirror1_x
        )
        ax.plot(x_seg, y_seg, z_seg, lw=2.2, color=c_main)

        if not inside_circular_aperture(y_hit, z_hit, mirror_radius):
            continue

        current_x, current_y, current_z = x_hit, y_hit, z_hit
        sign_x = +1

        # angle relative to cavity axis
        cos_theta_internal = abs(dx) / np.sqrt(dx**2 + dy**2 + dz**2)

        # detector focus position for this angle
        y_focus = f * (dy / abs(dx))
        z_focus = f * (dz / abs(dx))
        detector_r = np.sqrt(y_focus**2 + z_focus**2)

        # phase step between successive transmitted beams
        delta = (4 * np.pi * refractive_index * d * cos_theta_internal) / wavelength

        # collect transmitted intensities for this input ray
        transmitted_sum = 0.0

        for m in range(n_bounces):
            dx, dy, dz = direction_vector(theta, phi, sign_x=sign_x)
            target_plane = mirror2_x if sign_x > 0 else mirror1_x

            x_seg, y_seg, z_seg, x_hit, y_hit, z_hit = propagate_to_plane(
                current_x, current_y, current_z, dx, dy, dz, target_plane
            )

            ax.plot(x_seg, y_seg, z_seg, lw=2.0, color=c_main, alpha=0.8)

            if not inside_circular_aperture(y_hit, z_hit, mirror_radius):
                break

            # when ray reaches mirror 2, a transmitted beam leaves
            if sign_x > 0:
                # m-th transmitted beam intensity scaling
                # I_m ~ T^2 R^m
                I_m = (T**2) * (R**m)
                transmitted_sum += I_m

                # phase of m-th transmitted beam
                phase_m = m * delta

                # transmitted segment to lens
                dx_t, dy_t, dz_t = direction_vector(theta, phi, sign_x=+1)
                x_out, y_out, z_out, x_lens_hit, y_lens_hit, z_lens_hit = propagate_to_plane(
                    x_hit, y_hit, z_hit, dx_t, dy_t, dz_t, lens_x
                )

                # scale alpha by transmitted intensity
                alpha_ray = clip_alpha(I_m / (T**2))

                ax.plot(
                    x_out, y_out, z_out,
                    '--', lw=1.5 + 1.5*I_m/(T**2), color=c_main, alpha=alpha_ray
                )

                # after lens, all equal-angle rays go to same detector point
                s2 = np.linspace(0, 1, 250)
                x_after = lens_x + (screen_x - lens_x) * s2
                y_after = y_lens_hit + (y_focus - y_lens_hit) * s2
                z_after = z_lens_hit + (z_focus - z_lens_hit) * s2

                ax.plot(
                    x_after, y_after, z_after,
                    lw=1.2 + 1.3*I_m/(T**2), color=c_main, alpha=alpha_ray
                )

                # mark detector contribution
                ax.scatter(screen_x, y_focus, z_focus,
                           s=20 + 50*I_m/(T**2),
                           color=c_main, alpha=alpha_ray)

            current_x, current_y, current_z = x_hit, y_hit, z_hit
            sign_x *= -1

        # ideal Airy transmission for this angle
        F = 4*R/(1-R)**2
        I_airy = 1 / (1 + F * np.sin(delta/2)**2)

        # label once per input ray
        ax.text(
            screen_x + 0.15, y_focus, z_focus,
            f"r={detector_r:.2f}\nI={I_airy:.2f}",
            color=c_main, fontsize=8
        )

    # -------------------------
    # Labels / formatting
    # -------------------------
    ax.text(mirror1_x, 0, mirror_radius + 0.35, "Mirror 1")
    ax.text(mirror2_x, 0, mirror_radius + 0.35, "Mirror 2")
    ax.text(lens_x, 0, lens_radius + 0.35, "Lens")
    ax.text(screen_x, 0, detector_radius + 0.35, "Detector")

    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.set_zlabel("z")
    ax.set_title(
        f"3D Fabry–Perot interferometer\n"
        f"$\\lambda$ = {wavelength*1e9:.0f} nm,  R = {reflectivity:.2f}"
    )
    ax.set_box_aspect((12, 6, 6))
    plt.tight_layout()
    

points = [
    {"pos": (0.0, -2, -1), "theta": 0.15, "phi": 1.55},
]

plot_fp_3d_physical(points, n_bounces=8, wavelength=600e-9, reflectivity=0.85)

```

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_fp_3d_with_refracting_lens(points, n_bounces=8,
                                    wavelength=577e-9,          # cavity wavelength [m]
                                    reflectivity=0.85,          # mirror intensity reflectivity
                                    cavity_index=1.0,           # refractive index inside FP cavity
                                    lens_index=1.50,            # refractive index of lens glass
                                    air_index=1.0,
                                    mirror1_x=4.0, mirror2_x=6.0,
                                    lens_centre_x=9.0,
                                    lens_thickness=0.4,
                                    lens_radius_curvature=3.0,  # both surfaces
                                    lens_aperture_radius=2.0,
                                    screen_x=12.0,
                                    mirror_radius=2.5,
                                    detector_radius=3.0,
                                    detector_pixels=500,
                                    figsize_3d=(11, 8),
                                    figsize_2d=(8, 8)):
    """
    3D Fabry-Perot ray model with:
    - true cavity reflections
    - refraction through a bi-convex lens using Snell's law
    - 2D detector intensity map from Airy transmission

    Each item in points should be:
        {"pos": (x0, y0, z0), "theta": ..., "phi": ...}
    """

    # -------------------------
    # Basic quantities
    # -------------------------
    R = reflectivity
    T = 1 - R
    d = mirror2_x - mirror1_x

    # lens surface positions
    front_vertex_x = lens_centre_x - lens_thickness / 2
    back_vertex_x  = lens_centre_x + lens_thickness / 2

    # bi-convex lens: front bulges left, back bulges right
    front_centre = np.array([front_vertex_x + lens_radius_curvature, 0.0, 0.0])
    back_centre  = np.array([back_vertex_x  - lens_radius_curvature, 0.0, 0.0])

    # thin-lens focal length from lensmaker formula
    # symmetric biconvex: R1 = +Rc, R2 = -Rc
    R1 = lens_radius_curvature
    R2 = -lens_radius_curvature
    f_eff = 1.0 / ((lens_index / air_index - 1.0) *
                   (1.0 / R1 - 1.0 / R2 +
                    ((lens_index / air_index - 1.0) * lens_thickness) / ((lens_index / air_index) * R1 * R2)))

    # -------------------------
    # Helpers
    # -------------------------
    def unit(v):
        v = np.asarray(v, dtype=float)
        n = np.linalg.norm(v)
        if n == 0:
            raise ValueError("Zero-length vector encountered.")
        return v / n

    def direction_vector(theta, phi, sign_x=+1):
        dx = sign_x * abs(np.cos(theta) * np.sin(phi))
        dy = np.sin(theta) * np.sin(phi)
        dz = np.cos(phi)
        return unit(np.array([dx, dy, dz], dtype=float))

    def reflect(v, normal):
        normal = unit(normal)
        return unit(v - 2 * np.dot(v, normal) * normal)

    def refract(v, normal, n1, n2):
        """
        Vector Snell law.
        normal is the geometric surface normal.
        """
        v = unit(v)
        normal = unit(normal)

        if np.dot(v, normal) > 0:
            normal = -normal

        eta = n1 / n2
        cos_i = -np.dot(normal, v)
        k = 1 - eta**2 * (1 - cos_i**2)

        if k < 0:
            return None  # total internal reflection

        t = eta * v + (eta * cos_i - np.sqrt(k)) * normal
        return unit(t)

    def propagate_to_plane(p0, v, x_plane, npts=300):
        if abs(v[0]) < 1e-14:
            return None
        s_hit = (x_plane - p0[0]) / v[0]
        if s_hit <= 0:
            return None

        s = np.linspace(0, s_hit, npts)
        pts = p0[None, :] + s[:, None] * v[None, :]
        return pts, pts[-1]

    def line_to_sphere_intersection(p0, v, centre, radius):
        p0 = np.asarray(p0, dtype=float)
        v = unit(v)
        centre = np.asarray(centre, dtype=float)

        oc = p0 - centre
        b = 2 * np.dot(v, oc)
        c = np.dot(oc, oc) - radius**2
        disc = b**2 - 4 * c

        if disc < 0:
            return None

        root = np.sqrt(disc)
        t1 = (-b - root) / 2
        t2 = (-b + root) / 2

        valid = [t for t in (t1, t2) if t > 1e-9]
        if not valid:
            return None

        t_hit = min(valid)
        p_hit = p0 + t_hit * v
        return p_hit

    def inside_circle(y, z, radius):
        return y**2 + z**2 <= radius**2

    def airy_intensity(delta, R):
        F = 4 * R / (1 - R)**2
        return 1 / (1 + F * np.sin(delta / 2)**2)

    # -------------------------
    # 3D plot
    # -------------------------
    fig = plt.figure(figsize=figsize_3d)
    ax = fig.add_subplot(111, projection='3d')

    t = np.linspace(0, 2*np.pi, 300)

    # mirrors
    y_circle = mirror_radius * np.cos(t)
    z_circle = mirror_radius * np.sin(t)
    ax.plot(np.full_like(t, mirror1_x), y_circle, z_circle, color='black', lw=2.5)
    ax.plot(np.full_like(t, mirror2_x), y_circle, z_circle, color='black', lw=2.5)

    t_lens = np.linspace(0, 2*np.pi, 300)
    y_lens_outline = lens_aperture_radius * np.cos(t_lens)
    z_lens_outline = lens_aperture_radius * np.sin(t_lens)
    ax.plot(np.full_like(t_lens, lens_centre_x), y_lens_outline, z_lens_outline,
            color='black', lw=2)

    # detector outline
    yd = detector_radius * np.cos(t)
    zd = detector_radius * np.sin(t)
    ax.plot(np.full_like(t, screen_x), yd, zd, color='black', lw=2.5)

    ax.plot([0, screen_x], [0, 0], [0, 0], 'k--', alpha=0.25)

    colours = plt.rcParams['axes.prop_cycle'].by_key()['color']

    detector_hits = []

    for idx, ray in enumerate(points):
        c_main = colours[idx % len(colours)]

        p0 = np.array(ray["pos"], dtype=float)
        theta = ray["theta"]
        phi = ray["phi"]

        ax.scatter(*p0, s=45, color=c_main)

        # source -> mirror 1
        v = direction_vector(theta, phi, sign_x=+1)
        result = propagate_to_plane(p0, v, mirror1_x)
        if result is None:
            continue
        pts, p_hit = result
        ax.plot(pts[:, 0], pts[:, 1], pts[:, 2], lw=2.2, color=c_main)

        if not inside_circle(p_hit[1], p_hit[2], mirror_radius):
            continue

        current_p = p_hit.copy()
        current_v = v.copy()
        sign_x = +1

        # cavity angle to axis
        cos_theta_internal = abs(current_v[0]) / np.linalg.norm(current_v)
        delta = (4 * np.pi * cavity_index * d * cos_theta_internal) / wavelength

        for m in range(n_bounces):
            current_v = direction_vector(theta, phi, sign_x=sign_x)
            target_plane = mirror2_x if sign_x > 0 else mirror1_x

            result = propagate_to_plane(current_p, current_v, target_plane)
            if result is None:
                break
            pts, p_hit = result

            ax.plot(pts[:, 0], pts[:, 1], pts[:, 2], lw=2.0, color=c_main, alpha=0.8)

            if not inside_circle(p_hit[1], p_hit[2], mirror_radius):
                break

            # transmitted ray at mirror 2
            if sign_x > 0:
                I_m = (T**2) * (R**m)
                alpha_ray = max(0.08, min(1.0, I_m / (T**2)))

                # outgoing from cavity to front lens surface
                v_outside = direction_vector(theta, phi, sign_x=+1)
                p_front = line_to_sphere_intersection(p_hit, v_outside, front_centre, lens_radius_curvature)
                if p_front is not None and inside_circle(p_front[1], p_front[2], lens_aperture_radius):
                    ax.plot([p_hit[0], p_front[0]], [p_hit[1], p_front[1]], [p_hit[2], p_front[2]],
                            '--', lw=1.4, color=c_main, alpha=alpha_ray)

                    # refract air -> glass
                    n_front = unit(p_front - front_centre)
                    v_glass = refract(v_outside, n_front, air_index, lens_index)

                    if v_glass is not None:
                        # through lens to back surface
                        p_back = line_to_sphere_intersection(p_front + 1e-6 * v_glass, v_glass,
                                                             back_centre, lens_radius_curvature)
                        if p_back is not None and inside_circle(p_back[1], p_back[2], lens_aperture_radius):
                            ax.plot([p_front[0], p_back[0]], [p_front[1], p_back[1]], [p_front[2], p_back[2]],
                                    lw=1.4, color='tab:green', alpha=alpha_ray)

                            # refract glass -> air
                            n_back = unit(p_back - back_centre)
                            v_after = refract(v_glass, n_back, lens_index, air_index)

                            if v_after is not None:
                                result_det = propagate_to_plane(p_back + 1e-6 * v_after, v_after, screen_x)
                                if result_det is not None:
                                    pts_det, p_det = result_det
                                    ax.plot(pts_det[:, 0], pts_det[:, 1], pts_det[:, 2],
                                            lw=1.4, color='tab:red', alpha=alpha_ray)
                                    ax.scatter(*p_det, s=20 + 45 * I_m / (T**2),
                                               color='purple', alpha=alpha_ray)
                                    detector_hits.append((p_det, c_main))

            current_p = p_hit.copy()
            sign_x *= -1

        # label Airy intensity for this cavity angle
        I_airy = airy_intensity(delta, R)
        ax.text(screen_x + 0.1, 0.2 * idx, 0.2 * idx, f"I={I_airy:.2f}", color=c_main, fontsize=8)

    ax.text(mirror1_x, 0, mirror_radius + 0.3, "Mirror 1")
    ax.text(mirror2_x, 0, mirror_radius + 0.3, "Mirror 2")
    ax.text(lens_centre_x, 0, lens_aperture_radius + 0.3, f"Lens\nn={lens_index}")
    ax.text(screen_x, 0, detector_radius + 0.3, "Detector")

    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.set_zlabel("z")
    ax.set_title(
        f"3D Fabry–Perot with refracting lens\n"
        f"$\\lambda$ = {wavelength*1e9:.0f} nm,  R = {reflectivity:.2f},  n_lens = {lens_index:.2f}"
    )
    ax.set_box_aspect((12, 6, 6))
    plt.tight_layout()
    plt.show()

    # -------------------------
    # 2D detector intensity map
    # -------------------------
    y = np.linspace(-detector_radius, detector_radius, detector_pixels)
    z = np.linspace(-detector_radius, detector_radius, detector_pixels)
    Y, Z = np.meshgrid(y, z)
    rho = np.sqrt(Y**2 + Z**2)

    mask = rho <= detector_radius

    # lens maps angle -> radius approximately via rho = f_eff * tan(theta)
    theta_map = np.arctan(rho / abs(f_eff))

    delta_map = (4 * np.pi * cavity_index * d * np.cos(theta_map)) / wavelength
    I_map = airy_intensity(delta_map, R)
    I_map[~mask] = np.nan

    plt.figure(figsize=figsize_2d)
    plt.imshow(
        I_map,
        extent=[y.min(), y.max(), z.min(), z.max()],
        origin='lower',
        cmap='inferno'
    )
    plt.xlabel("y on detector")
    plt.ylabel("z on detector")
    plt.title(
        f"Fabry–Perot detector intensity\n"
        f"$f_\\mathrm{{lens}}$ = {f_eff:.2f},  n_lens = {lens_index:.2f}"
    )
    plt.gca().set_aspect('equal')
    plt.colorbar(label="Normalised intensity")

    # overlay traced detector hits
    if detector_hits:
        hit_y = [p[1] for p, _ in detector_hits]
        hit_z = [p[2] for p, _ in detector_hits]
        hit_c = [c for _, c in detector_hits]
        plt.scatter(hit_y, hit_z, c=hit_c, s=25, edgecolors='white', linewidths=0.5)

    plt.show()

    return I_map, f_eff


# example
points = [
    {"pos": (0.0, -2.0, -1.0), "theta": 0.15, "phi": 1.55},
]

I_map, f_eff = plot_fp_3d_with_refracting_lens(
    points,
    n_bounces=8,
    wavelength=600e-9,
    reflectivity=0.85,
    cavity_index=2.0,   # Fabry-Perot cavity refractive index
    lens_index=1.50     # actual lens refractive index
)
```
